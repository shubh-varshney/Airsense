```python
import pandas as pd
from pathlib import Path

data_dir = Path("../data/raw")

files = [
    "pm25_measurements.csv",
    "pm25_measurements2019.csv",
    "pm25_measurements2020_1.csv",
    "pm25_measurements2020_2.csv",
    "pm25_measurements2021_1.csv",
    "pm25_measurements2021_2.csv",
    "pm25_measurements2022_1.csv",
    "pm25_measurements2022_2.csv"
]

frames = []

for file in files:
    df = pd.read_csv(data_dir / file)
    frames.append(df)
    print(file, df.shape)

combined = pd.concat(frames, ignore_index=True)

print("\nCombined shape:", combined.shape)

combined.to_csv(
    data_dir / "pm25_all_years.csv",
    index=False
)
```
